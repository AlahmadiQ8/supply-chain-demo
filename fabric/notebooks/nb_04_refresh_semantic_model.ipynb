# 04 · Refresh the semantic model
Reframes the Direct Lake semantic model **Supply Chain Trips** so it points at the latest
Gold Delta table versions right after the Gold notebook finishes. Uses Semantic Link (`sempy`).

```python
semantic_model = "Supply Chain Trips"
```

```python
import time

import sempy.fabric as fabric

if semantic_model not in set(fabric.list_datasets()["Dataset Name"]):
    notebookutils.notebook.exit(f"Semantic model '{semantic_model}' not found - nothing to refresh")

request_id = fabric.refresh_dataset(semantic_model, refresh_type="full")
for _ in range(60):
    details = fabric.get_refresh_execution_details(semantic_model, request_id)
    if details.status not in ("Unknown", "NotStarted", "InProgress"):
        break
    time.sleep(10)
print(f"Refresh status: {details.status}")
assert details.status == "Completed", f"Refresh failed: {details.messages}"
```
